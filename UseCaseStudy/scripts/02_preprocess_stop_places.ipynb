## 02 Preprocess Stop Places

```python
# import
import geopandas as gpd

# inread data
tarif_area_schwyz = gpd.read_file('../data/tarif-area/tarifverbund_schwyz_merged.shp')
tarif_area_vagabond = gpd.read_file('../data/tarif-area/vagabond_merged.shp')
stop_places = gpd.read_file(
    '../data/haltestellen-oev_2056_de.gpkg/HaltestellenOeV.gpkg', 
    layer = 'Betriebspunkt',
    columns = ['Name', 'Nummer', 'geometry']
)
stop_places.head()
```

```python
if stop_places.crs != tarif_area_schwyz.crs:
    tarif_area_schwyz = tarif_area_schwyz.to_crs(stop_places.crs)

if stop_places.crs != tarif_area_vagabond.crs:
    tarif_area_vagabond = tarif_area_vagabond.to_crs(stop_places.crs)
    
print(stop_places.crs)
assert stop_places.crs == tarif_area_schwyz.crs
assert stop_places.crs == tarif_area_vagabond.crs

# filter stop_places, only keep those within tarif area
stop_places_schwyz = gpd.clip(stop_places, tarif_area_schwyz)
stop_places_vagabond = gpd.clip(stop_places, tarif_area_vagabond)

# Check the result
print(f"Original stops: {len(stop_places)}")
print(f"Stops in Schwyz: {len(stop_places_schwyz)}")
print(f"Stops in Vagabond: {len(stop_places_vagabond)}")

stop_places.to_file('../data/stop-places-schweiz/stop-places-schweiz.shp')
stop_places_schwyz.to_file('../data/stop-places-schwyz/stop-places-schwyz.shp')
stop_places_vagabond.to_file('../data/stop-places-vagabond/stop-places-vagabond.shp')
```
